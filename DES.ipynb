## Step1: Initial permutation

```python
import numpy as np 
b=np.arange(0,64,1)
print(b, b.shape, sep='\n')
```

```python
IP_matrix=np.full(shape=(8,8), fill_value=-1)
k=57
for i in range(4):
    temp=k
    for j in range(8):
        IP_matrix[i][j]=b[temp]
        temp-=8
    k+=2
k=56
for i in range(4,8,1):
    temp=k
    for j in range(8):
        IP_matrix[i][j]=b[temp]
        temp-=8
    k+=2
print(IP_matrix)

```

## Step2: Expansion

```python
left_matrix=IP_matrix[0:4:1,]
right_matrix=IP_matrix[4:8:1,]
left_matrix=left_matrix.reshape(8,4)
right_matrix=right_matrix.reshape(8,4)
print(left_matrix, right_matrix, sep='\n')
```

```python
E0 =np.array([
    31,  0,  1,  2,  3,  4,
     3,  4,  5,  6,  7,  8,
     7,  8,  9, 10, 11, 12,
    11, 12, 13, 14, 15, 16,
    15, 16, 17, 18, 19, 20,
    19, 20, 21, 22, 23, 24,
    23, 24, 25, 26, 27, 28,
    27, 28, 29, 30, 31,  0
])

left_expanded =left_matrix.flatten()[E0]  
left_expanded=left_expanded.reshape(8,-1)
right_matrix=right_matrix.flatten()[E0]
right_matrix=right_matrix.reshape(8,-1)
print(left_expanded, right_matrix, sep='\n')
```
